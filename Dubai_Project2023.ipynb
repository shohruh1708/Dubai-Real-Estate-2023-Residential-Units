```python
import pandas as pd 

df = pd.read_csv("Transactions.csv")


pd.set_option('display.max_columns', None)

pd.set_option('display.max_rows', None)
```

```python
print(len(df[df.duplicated()]))

# As can be seen there are no duplicates exists
```

```python
# We are going to drop all the column that end with '_ar' since english version of it already exists on the table

column_to_drop = [col for col in df.columns if col.endswith('_ar') ]

column_to_drop

df = df.drop(columns=column_to_drop)

df.head()
```

```python
# Checking the datatypes

df.info()
```

```python
# Now as can be seen, that  instance_date is on string type
# we need to convert to datetime so we are able to work on specific dates

df['date'] = pd.to_datetime(df['instance_date'], format='%d-%m-%Y', errors='coerce')

```

```python
len(df[df['date'].isnull()])
```

```python
df.isnull().sum()
```

```python
# As can been seen there are 3 types of transactions: ['Sales', 'Mortgages', 'Gifts']

df['trans_group_en'].unique()
```

```python

# Here i am selecting year by 2023 and transactions by Sales and Mortgages, not including Gifts. 

result = df[(df['date'] >='2023-01-01')
            &
            (df['date'] <='2023-12-31')]

result = result[df['trans_group_en'] != 'Gifts']

result


```
